# Strategic Audit of the PCV Program in Guatemala 
### First Notebook: Project Overview, Data Framework, and Analytical Design  

This notebook begins a strategic audit of the PCV (Volunteer Cooperation Program) in Guatemala in a context where most available information is qualitative, fragmented, and not organized into a clear program structure. Deliverables are non-standardized, structured data is limited, and many activities are documented across separate sources, making direct analysis difficult.

For this reason, the first step of the audit is to reconstruct and organize the program before analyzing it.

This notebook builds the main analytical framework of the project by organizing activities into capacity areas, operational components (such as diagnostics, policies, tools, and training activities), and project-level analytical units.

At the same time, a series of exploratory proxy indicators is introduced to approximate important organizational and operational dimensions where direct measures are not available. These indicators are not meant to provide exact measurements, but rather to support comparison and interpretation across projects and activities.

The objective of this notebook is not yet to evaluate performance or identify organizational patterns, but to create a consistent analytical structure that will support the following stages of descriptive analysis, exploratory modeling, and strategic interpretation.

Overall, this notebook adopts an exploratory reconstruction-based approach to transform fragmented qualitative program information into a structured analytical system adapted to a low-data international development context.

## 1.1-Context of the Assignment

The PCV program in Guatemala was implemented through long-term partnerships with local organizations, structured around progressive capacity strengthening processes. As reflected in the *camino del cambio*, partner organizations followed multi-year trajectories combining diagnostics, capacity strengthening plans, and a wide range of activities including trainings, policy development, tools, and participatory events.

These interventions generated diverse outputs and perceived impacts across key domains such as project management, gender equality, environmental practices, and organizational development. However, this information remains largely qualitative, dispersed, and unevenly documented, with varying levels of continuity due to factors such as staff and volunteer turnover and implementation gaps.

This context highlights the need for a structured analytical approach capable of synthesizing fragmented program information into a coherent framework for assessment. 

**However, this framework is exploratory and interpretative in nature. It is designed to structure heterogeneous qualitative information into an analyzable format, rather than to provide statistically validated measurements or causal inference.**

## 1.2-Analytical Approach

Given the lack of structured data and standardized program design, the analysis follows a reconstruction approach aimed at organizing program activities into a coherent and analyzable structure.

This approach is based on the assumption that capacity strengthening interventions rely on recurring building blocks, such as diagnostics, development of frameworks or tools, and capacity-building activities.

The analysis therefore consists of reorganizing all identified activities according to this logic, in order to create a consistent representation of the program across different partners and interventions.

This structured representation provides a foundation for further analysis in subsequent sections.

## 1.4-Proxy Indicator System

To enable structured analysis in the absence of standardized metrics, a system of proxy indicators was developed to translate qualitative and heterogeneous program data into analyzable dimensions.

This system is based on a set of **scoring scales from 1 to 5**, designed to capture specific and non-overlapping dimensions of program implementation. These scores are assigned at two distinct levels:

* **Deliverable level**, where indicators capture the operational characteristics of production
* **Project level**, where indicators reflect quality, adoption, outcomes, and contextual conditions

At the deliverable level, scores are used to construct a **Complexity Index**, which captures the operational intensity of production through dimensions such as effort, internal mobilization, external engagement, and methodological novelty.

At the project level, scores are aggregated into several composite indices:

* **Quality Index**, reflecting the intrinsic quality and technical robustness of project outputs based on the expected standards of each project type.
* **Adoption Index**, capturing alignment with partner needs, satisfaction, usage, and sustainability
* **Transformation Signal Index**, indicating the extent of change in practices and institutionalization
* **Outcome Signal Index**, reflecting perceived effects on organizational performance, beneficiaries, economic value, and network influence
* **Context Index**, representing organizational and relational conditions affecting implementation
* **Risk Index**, accounting for internal instability and external volatility

These composite indices combine several underlying scores into broader analytical dimensions through heuristic aggregation (a rule-based approach that combines indicators using analytical judgment rather than statistical weighting). Their purpose is to create a consistent and comparable framework across different types of interventions while keeping the results easy to interpret.

## 1.5-Data Completion Methodology

The dataset was constructed through a multi-source completion process combining qualitative and documentary inputs. Information was primarily derived from the *camino del cambio*, the final program report, and additional documentation provided by the country representative and other cooperants involved in the program.

This information was complemented by a systematic review of available internal documents produced throughout the program cycle, allowing for cross-validation and triangulation of key elements.

However, the dataset is characterized by the absence of structured questionnaires and standardized measurement tools. In several cases, available documentation did not provide sufficiently precise or complete information, requiring the use of informed estimations to complete missing values.

Given these constraints, a degree of uncertainty is associated with certain data points. This uncertainty is acknowledged and will be explicitly reflected in the interpretation of results and the formulation of recommendations.

**Summary - Limitations of the Approach**

| Limitation                     | Description                                                                             | Implication                                                             |
| ------------------------------ | --------------------------------------------------------------------------------------- | ----------------------------------------------------------------------- |
| Lack of standardized data      | No structured questionnaires or unified monitoring system were available                | Limits comparability and precision across variables                     |
| Qualitative reconstruction     | Several variables were derived from qualitative interpretation and informed estimations | Introduces a degree of subjectivity in the dataset                      |
| Proxy-based measurement        | Indicators are constructed rather than directly observed                                | Results should be interpreted as analytical signals, not exact measures |
| Aggregation effects            | Composite indices combine heterogeneous dimensions                                      | May mask internal variability within each index                         |
| Partial documentation coverage | Analysis relies on available reports and stakeholder inputs                             | Some activities or perspectives may be underrepresented                 |

Despite these limitations, the framework provides a structured starting point for analyzing the program in a context of fragmented and limited data. The objective is not to claim perfect measurement, but to progressively build more coherent and objective analytical tools that can support comparative interpretation and future refinement of the system.
