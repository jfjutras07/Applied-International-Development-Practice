# Designing and Implementing an Excel-Based Project Prioritization Tool  
## Analysis, Tool Design, and Implementation

This project presents a simple decision-support tool designed to help organizations prioritize projects within a portfolio. The tool allows users to register proposed projects, evaluate them using a structured prioritization grid, and visualize the overall portfolio through a bubble chart that highlights the relative importance and characteristics of each initiative.

The objective of this notebook is not to develop complex optimization models, but to provide a clear and transparent framework for comparing projects and supporting strategic decision-making. The proposed approach relies on structured evaluation criteria, scoring mechanisms, and a visual portfolio representation that helps decision-makers quickly identify high-priority initiatives and balance competing demands.

This notebook documents the logic behind the prioritization grid, the evaluation criteria used to score projects, and the graphical representation used to visualize the project portfolio. It illustrates how a lightweight Excel-based system can support project selection and portfolio discussions without requiring specialized portfolio management software.

## 1.1-Context of the Assignment

This work was carried out as part of my first six-month assignment with SUCO in Guatemala, where I served as a Project Management Collaborator based in Antigua.

Within the regional operational context, SUCO and its partner organizations regularly face the challenge of selecting among multiple potential initiatives while operating under limited financial and human resources. Project proposals may originate from internal teams, partner organizations, or external funding opportunities, creating the need for structured mechanisms to compare and prioritize initiatives.

While project selection discussions were already taking place informally, no structured framework existed to systematically compare proposals based on strategic relevance, operational feasibility, and expected impact. As a result, prioritization decisions were often based on partial information or informal deliberations.

Within this context, my mandate included supporting internal project management processes and contributing to practical tools that could help structure project selection discussions and improve transparency in decision-making.

## 1.2-Initial Needs Analysis

The initial needs analysis was conducted through discussions with SUCO staff, partners, and a review of existing project planning practices. These exchanges aimed to better understand how project proposals were currently identified, evaluated, and selected within the organization.

The analysis highlighted several recurring challenges. Project proposals varied significantly in scope, scale, and level of preparation. In addition, different stakeholders tended to prioritize different dimensions, such as potential impact, alignment with organizational priorities, or the availability of funding opportunities.

Another challenge identified was the absence of a consolidated portfolio view of potential initiatives. Without a structured overview, it was difficult to compare projects or to visualize how they related to one another in terms of expected impact, implementation complexity, and resource requirements.

An existing Excel-based tool was already being used to support elements of project evaluation. However, this tool presented several limitations. The evaluation criteria were primarily designed for strategic planning purposes and focused mainly on the organizational capacity of partner organizations. In addition, the criteria were not weighted, meaning that all dimensions were implicitly treated as equally important regardless of their strategic relevance.

The tool also relied entirely on manual calculations and did not include automated scoring or visual representations that could facilitate interpretation of the results. As a result, while the tool provided a useful starting point, it did not fully support structured prioritization discussions or portfolio-level analysis.

The analysis therefore pointed to the need for a revised tool capable of:

- Structuring project information in a consistent and comparable way  
- Introducing a transparent weighting mechanism for evaluation criteria  
- Supporting the comparison of multiple project proposals  
- Facilitating prioritization discussions among decision-makers  
- Providing a visual overview of the project portfolio  

Given the technological environment and the digital literacy of users, Excel was identified as the most appropriate platform to develop such a tool. Building on an existing familiar platform also helped ensure that the proposed solution would remain accessible and easy to adopt.

## 1.3-Tool Design

### Initial Design Phase

Following the initial analysis, the development of the tool focused on translating the project evaluation process into a structured and easy-to-use Excel framework.

At the time of the analysis, project prioritization discussions already took place between SUCO and partners. However, these discussions were generally informal and relied on partial information or individual judgment rather than on a shared and structured evaluation framework.

The initial design effort therefore focused on creating a simple prioritization mechanism capable of structuring project comparisons while remaining accessible to users with varying levels of technical familiarity.

Given the technological environment and the digital literacy of users, Excel was identified as the most appropriate platform for the development of the tool. Excel is widely used within the organization and by partner institutions, making it a practical choice for building a lightweight decision-support system without introducing additional technological barriers.

The objective was not to create a complex portfolio management system, but rather to provide a transparent framework capable of supporting structured discussions around project prioritization and portfolio balance.

### Identified Needs and Design Trade-Offs

During consultations with SUCO staff and partners, several needs were identified regarding the prioritization of project proposals. These needs included the ability to compare initiatives, structure evaluation criteria, and visualize the project portfolio.

While it would have been possible to incorporate additional project management elements, the design focused on a limited and coherent scope specifically dedicated to project prioritization, to preserve usability and facilitate adoption.

| Area | Identified Needs | Design Decision |
|-----|-----|-----|
| Centralized Project Registry | Collect all proposed projects consistently, with structured information to support comparison | **Fully integrated** as the core component of the tool; all projects recorded in a single sheet (Lista de proyectos) |
| Weighted Evaluation Criteria | Ability to assign different importance to criteria to reflect strategic priorities | **Fully integrated**; each criterion can be weighted, scores automatically aggregated |
| Automated Scoring | Minimize manual calculations and ensure consistency | **Fully integrated**; scoring (1-5-10) automated with formulas |
| Portfolio Visualization | Visual overview of all projects to support decision-making | **Fully integrated** as a bubble chart; X = Impact, Y = Effort, Bubble size = Total Score |
| User-friendly Interface | Intuitive form-based interface for evaluating projects | **Fully integrated** via Priorización worksheet with input forms and validation |
| Deliverables & Participants | Tracking outputs and stakeholders for reporting purposes | **Excluded** |
| Dashboard and Key Indicators | Consolidated monitoring indicators | **Excluded** |

### Tool Structure and Architecture

User stories and use cases were identified during the analysis phase to clarify how the tool could support project prioritization and portfolio discussions. However, these elements are not presented explicitly in this notebook in order to keep the documentation concise and focused on the practical structure of the tool.

Instead, this section presents the main components of the Excel tool and explains how they interact to support the prioritization process.

The tool is organized around three main worksheets that structure the evaluation process and the visualization of the project portfolio.

**Presentation**

The tool includes a dedicated sheet that provides an overview of its structure and main components. This sheet briefly explains the purpose of each worksheet and how they interact within the overall prioritization process.

It serves as an entry point for users and provides simple instructions for navigating the tool. By presenting the overall logic of the workbook, this sheet helps ensure that both new and experienced users can quickly understand how the tool should be used.

**Project Evaluation Form**

This worksheet constitutes the main interface of the prioritization tool. Users identify the project to be analyzed and evaluate it according to a predefined set of prioritization criteria. Each criterion is scored using a standardized **1–5–10 scale**, allowing evaluators to express different levels of importance or intensity while maintaining consistency across evaluations.

The criteria used to evaluate each project are:

- Strategic Alignment  
- Business Value  
- Urgency  
- Number of Stakeholders Impacted  
- Efforts Required  
- Technical Complexity  
- Risk Level  
- Dependencies  

These criteria were selected to capture both the **expected impact of the project** and the **level of effort or complexity associated with its implementation**. This balanced approach helps decision-makers consider not only the potential benefits of an initiative but also the resources and constraints involved in its execution.

The tool automatically calculates the aggregated score of the project based on the selected criteria and the defined weighting structure. This automation reduces manual calculations and ensures consistent evaluation across projects.

All user inputs are limited to clearly identified input fields, while the remaining values are calculated automatically using formulas embedded in the worksheet. This structure helps reduce data entry errors and improves the reliability of the prioritization results.

### Project Prioritization Matrix

The sheet provides a graphical overview of all evaluated projects within the portfolio. Projects appear in a **bubble chart** that allows users to quickly compare initiatives across three dimensions:

- **X-axis:** Project impact  
- **Y-axis:** Implementation effort  
- **Bubble size:** Total prioritization score

Each project evaluated in the **Priorización** worksheet is automatically represented in this chart. This visualization provides a consolidated view of the project portfolio and helps decision-makers quickly identify which initiatives combine strong potential impact with manageable implementation effort.

To support interpretation of the chart, a simple **matrix is included in the sheet to explain the meaning of each quadrant**. This matrix helps users understand how projects positioned in different areas of the chart should typically be interpreted in the prioritization process.

For example, projects located in the **high-impact / low-effort quadrant** may represent priority opportunities, while projects with **low impact and high effort** may require further justification before being pursued.

To facilitate repeated use of the tool, the sheet also includes a **reset button** that clears the matrix and associated data. This feature allows users to quickly start a **new prioritization exercise** without manually removing previous inputs, ensuring that the tool can be reused easily during different planning or decision-making sessions.

## 1.4 Tool Implementation and User Support

The implementation of the project prioritization tool was carried out with partners as part of efforts to strengthen project management and decision-support practices.

The Excel workbook was first tested using a small number of existing project proposals in order to validate the scoring logic, the weighting mechanism, and the graphical portfolio visualization. This testing phase helped ensure that the calculations and visual representations accurately reflected the intended prioritization framework.

Once validated, the tool was shared with relevant team members involved in project planning and proposal discussions. A brief explanation of the tool’s structure and evaluation process was provided to facilitate its use during internal prioritization discussions.

Given the simplicity of the tool and its reliance on standard Excel functionalities, no extensive training was required. The design intentionally prioritized usability and transparency in order to allow users to quickly understand the evaluation process and integrate the tool into existing decision-making practices.

## 1.5-Post-Implementation Development and Future Enhancements

Following the initial implementation of the prioritization tool, several potential improvements were identified that could further strengthen its usefulness for project portfolio discussions.

Future developments may include refining the prioritization criteria and the weighting structure as organizational priorities evolve. Adjustments could also be introduced to better reflect the specific operational realities and strategic planning frameworks used by partner organizations, their donors, and SUCO.

Additional improvements could involve expanding the descriptive information associated with project proposals in order to provide greater context during evaluation and prioritization discussions.

Finally, the tool could be progressively integrated with other project management and monitoring mechanisms used by partner organizations. Such integration would help ensure stronger coherence between project selection processes and the follow-up of approved initiatives.

These potential enhancements are envisioned as incremental adjustments rather than major structural changes, ensuring that the tool remains simple, accessible, and easy to use.

## 1.6-Key Lessons Learned

One of the main lessons from this initiative is that effective decision-support tools do not necessarily require complex systems. In many organizational contexts, a simple and transparent framework can significantly improve the structure and quality of strategic discussions.

The development of the prioritization tool demonstrated the importance of balancing analytical rigor with usability. By limiting the number of evaluation criteria and relying on a clear scoring mechanism, the tool remains accessible while still supporting structured comparisons between project proposals.

Another key insight concerns the value of visual representations in decision-making. The use of a bubble chart to map projects according to impact, effort, and overall score provides an intuitive overview of the project portfolio and helps facilitate collective discussions among decision-makers.

Finally, the experience highlighted that prioritization tools are most useful when they are integrated into existing planning and coordination processes rather than used as standalone analytical instruments. Anchoring the tool in real organizational practices increases the likelihood that it will support more transparent and consistent project selection decisions over time.
