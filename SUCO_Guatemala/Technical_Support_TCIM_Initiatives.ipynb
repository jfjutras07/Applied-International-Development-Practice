# Technical Support for Traditional, Complementary and Integrative Medicine (TCIM) Initiatives
### Technical Support, Health Promotion, and Good-Practice Review

This project documents technical support provided to partner organizations in Guatemala during my six-month assignment with SUCO. The work combined support for health-promotion activities, technical review of product deployment approaches, and review of health-related materials and products against good practices.

The support covered three complementary areas. In Sololá, I supported the organization of a health-promotion day and contributed to ensuring that content presented by health professionals (nurses and nutritionists), including nutrition-related information and information on locally cultivated medicinal plants, was aligned with good practices. I also provided technical support for the review of methods used to deploy products developed through the amaranth value chain. With a second partner, I supported the review of a medicinal plants guide and related products intended for sale, with attention to alignment with good practices.

The objective of this work was not to replace the technical expertise of health professionals or partner organizations, but to provide complementary technical support that could strengthen the quality, consistency, and practical application of TCIM-related activities and materials.

**Author**: J-F Jutras  
**Date**: February–August 2026  
**Scope**: Technical support to partner organizations on TCIM-related health promotion, product deployment, and review of health-related materials and products against good practices

## 1.1-Context of the Assignment

This work was carried out as part of my six-month assignment with SUCO in Guatemala, where I served as a Project Management Associate. Within this mandate, I provided complementary technical support to partner organizations involved in health promotion, medicinal plants, natural health products, and amaranth-related activities.

One of the interventions took place in Sololá in April 2026, where a public health day was organized for members of the local community. The activity brought together several information and service areas, including a nutrition information booth, a booth focused on improving the nutritional value of traditional foods through the use of superfoods and natural health products, health assessments provided by physicians and nurses, and a medicinal plants booth combining information and product sales. The activity required both organizational support and technical review to ensure that the information presented to community members was consistent with good practices and could be communicated in an accessible and appropriate way.

A second intervention focused on an amaranth-related initiative involving local production and the development of products intended to support both household nutrition and income generation. The initiative involved women from several communities producing amaranth for family consumption and economic sustainability. Within this context, the support focused on reviewing the pathway from production through product development, presentation, and sales in order to identify opportunities to improve how products were developed, presented, and brought to users and markets.

A third intervention involved the review of a medicinal plants guide and related products developed by another partner organization. The review considered the alignment of the guide with good practices, the relationship between medicinal plant cultivation and seasonal conditions, the preparation of related products, and the accuracy of information describing their effects. The work also considered practical approaches for monitoring sales, discussing user satisfaction, and exploring ways to observe potential outcomes associated with the use and dissemination of these products.

Across these interventions, the technical support was intended to complement the existing expertise of health professionals and partner organizations. The work therefore focused on strengthening the quality and consistency of health-related information, improving the presentation and practical use of TCIM-related products and materials, and introducing simple approaches to support monitoring and learning.

## 1.2-Initial Needs Analysis

The initial needs analysis was conducted to identify the main technical, organizational, communication, and monitoring needs associated with the three partner initiatives. The analysis focused on understanding existing practices, identifying gaps or areas requiring clarification, and determining where complementary technical support could add value.

The needs identified across the three interventions related to five main areas: quality and responsible communication of health-related information, adaptation to local context, product development and deployment, alignment between information and existing practices, and simple approaches to monitoring and learning.

**1.2.1-Health-Promotion Day in Sololá**

The initial needs analysis for the health-promotion day focused on the technical quality of the information and its accessibility to the community. The planned activity included nutrition information, preparation of foods using superfoods and natural health products, health assessments conducted by physicians and nurses, and a medicinal plants booth providing information and products for sale.

A key consideration was the linguistic and educational context of the community. Many participants primarily spoke Kaqchikel rather than Spanish, requiring translators to facilitate communication between health professionals and community members. Many participants also had limited formal education. These factors created a need to ensure that technical health information could be communicated clearly without relying heavily on specialized terminology or written materials.

A second need concerned the consistency of information presented across the different components of the activity. Several health professionals and thematic areas were involved, requiring coordination to ensure that information was complementary, technically appropriate, and understandable to participants.

The needs identified therefore included:

| **Area** | **Identified Needs** |
| --- | --- |
| **Health Information** | Ensure that information presented to participants was technically appropriate and consistent with good practices. |
| **Language Accessibility** | Ensure that information could be communicated effectively to participants who primarily spoke Kaqchikel. |
| **Community Accessibility** | Adapt health information to participants with different levels of formal education. |
| **Nutrition and Natural Health Products** | Ensure appropriate and responsible presentation of nutritional and health-related information. |
| **Medicinal Plants** | Provide clear and responsible information on locally cultivated medicinal plants and related products. |
| **Activity Coordination** | Maintain consistency and complementarity between the different thematic areas and health professionals involved. |

The analysis therefore identified a need for complementary technical and organizational support during preparation of the activity, with particular attention to technical accuracy, responsible communication, linguistic accessibility, and adaptation to the community context.

**1.2.2-Technical Support for the Amaranth Product Pathway**

The initial needs analysis for the amaranth-related initiative focused on understanding how products developed from locally produced amaranth moved from production to product development, presentation, and sales. The initiative had both nutritional and economic dimensions, with women from several communities producing amaranth for household consumption and income generation.

The analysis identified a need to better understand the existing product pathway and how the characteristics and potential value of amaranth were reflected in the products developed and presented to potential users.

A second need concerned the transition between product development, presentation, and sales. Understanding how products were communicated and brought to potential users was necessary to identify practical opportunities for improvement.

The needs identified therefore included:

| **Area** | **Identified Needs** |
| --- | --- |
| **Product Pathway** | Clarify how products moved from local production through development, presentation, and sales. |
| **Product Development** | Examine how the characteristics of amaranth were reflected in the products developed. |
| **Product Presentation** | Identify opportunities to improve how products were presented to potential users. |
| **Product Communication** | Ensure that information about the products was clear and appropriate. |
| **Sales and Deployment** | Identify opportunities to improve how products reached potential users and markets. |

The analysis therefore identified opportunities to better understand and optimize the existing product pathway, particularly the links between product development, presentation, communication, deployment, and sales.


**1.2.3-Review of the Medicinal Plants Guide and Related Products**

The initial needs analysis for the medicinal plants initiative focused on an existing guide and related products developed by the partner organization. The review considered the quality and consistency of the information provided to users, as well as the conditions under which medicinal plants were cultivated, prepared, presented, and sold.

A first area of attention was the content of the medicinal plants guide. Information on individual plants, including Pericón, Ajenjo, Lavanda, Orégano, Romero, Té de limón, Albahaca, and Tomillo, required review to identify information requiring correction, clarification, or more appropriate presentation, particularly when describing traditional uses or potential effects.

A second area concerned the relationship between the guide and the partner's medicinal plant gardens. The analysis considered whether information about cultivation and availability was consistent with seasonal conditions and whether the guide could better support the practical management and use of the plants.

The analysis also identified a need to clarify information concerning the preparation and intended uses of products derived from medicinal plants. Particular attention was required to distinguish traditional uses from health-related claims requiring stronger evidence.

Finally, there was a need for simple approaches to monitor products after they were made available for sale, including sales, user satisfaction, and observations that could support future learning.

The needs identified therefore included:

| **Area** | **Identified Needs** |
| --- | --- |
| **Guide Content** | Correct and clarify information provided about medicinal plants. |
| **Good Practices** | Ensure that information and product descriptions were presented responsibly. |
| **Cultivation** | Better connect the guide with seasonal medicinal plant cultivation and availability. |
| **Product Preparation** | Clarify information related to the preparation and intended use of medicinal plant products. |
| **Product Sales** | Establish simple approaches to monitor products once they were made available for sale. |
| **User Feedback** | Create opportunities to obtain basic information on users' experiences and satisfaction. |
| **Potential Effects** | Explore appropriate ways to document observations without making unsupported health claims. |

The analysis therefore identified a need to strengthen the guide and related products through clearer information, better alignment with cultivation and preparation practices, and simple approaches to monitoring sales, user feedback, and potential outcomes.

## 1.3-Technical Support and Intervention Design

The intervention design was developed directly from the needs identified during the initial analysis. When relevant documentation was available, the review was informed by technical resources from major international organizations and reference bodies working in traditional, complementary, and integrative medicine (TCIM), including the World Health Organization (WHO), the Pan American Health Organization (PAHO/WHO), and other recognized technical and professional reference sources. These resources were used to support the review of health-related information, good practices, and the responsible presentation of TCIM-related content.

For the purposes of this project summary, the specific reference documents and technical sources are not described in detail. The approach was to use available authoritative guidance where applicable while recognizing that not all information or practices encountered in the partner organizations' activities had a sufficiently established evidence base.

Where the available evidence or guidance was not conclusive, the relevant elements were discussed with the professionals involved through working meetings and group review sessions. These discussions were used to clarify the information, assess its applicability to the local context, and, where necessary, revise the content or proposed approach. The objective was not to establish clinical efficacy independently, but to support accurate, responsible, and context-appropriate technical information.

The intervention design for the health-promotion day focused on the quality, accessibility, and coordination of the information to be presented to community members. The identified needs were addressed through content review, reference to relevant technical guidance, stakeholder coordination, and working discussions with the professionals involved.

**1.3.2-Amaranth Product Pathway**

The intervention design for the amaranth initiative focused on understanding the existing pathway from production through product development, presentation, deployment, and sales. Different analytical approaches were used to examine the product pathway and the way products were presented and perceived by potential users.

--- | --- | ---
Product Pathway | Review of partner documentation, process mapping, and flow analysis | Examine the sequence of activities from production through product development, deployment, and sales and identify potential points for improvement

The combination of documentary review, visual process mapping, flow analysis, group discussions, semi-structured interviews, and qualitative data provided a basis for identifying practical opportunities to optimize the existing product pathway and strengthen product presentation and deployment.



**1.3.3-Medicinal Plants Guide and Related Products**

The intervention design for the medicinal plants initiative was based on a structured review of the existing guide, the medicinal plant cultivation context, and the products associated with the initiative. Available TCIM technical references were consulted where relevant, particularly when reviewing information concerning traditional uses, preparation, and responsible presentation of health-related information.

Need | Analysis or Tool Used | Purpose
--- | --- | ---
Guide Content | Document review using available technical references | Identify information requiring correction, clarification, or improved presentation
Good Practices | Review of health-related information and discussion with professionals | Identify statements requiring more careful or responsible presentation
Cultivation | Review of the relationship between the guide and seasonal cultivation | Identify opportunities to better align the guide with the availability and cultivation of medicinal plants
Product Preparation | Review of preparation information and available technical references | Examine whether information concerning product preparation was sufficiently clear and appropriate
Product Sales | Review of available sales information | Establish a basis for monitoring the products made available for sale

The user feedback tool was intentionally simple and designed to obtain basic information during existing interactions with community members, rather than create a separate or burdensome data-collection process. The information collected provided an initial source of descriptive data to support future learning and improvement.

When the available evidence was limited or inconclusive, the relevant information was reviewed with the professionals involved and modified where appropriate. This process helped maintain a distinction between established information, traditional uses, and areas where stronger evidence would be required before making more definitive health-related claims.

## 1.5-Key Lessons Learned

The assignment highlighted several practical lessons about providing technical support for TCIM-related activities and products in community and partner settings.

| **Key Lesson** | **Learning from the Assignment** |
| --- | --- |
| **Evidence and Good Practices Must Guide Technical Support** | TCIM-related information and products should be reviewed against the best available evidence and relevant technical guidance. When evidence is limited or inconclusive, information should be presented carefully and discussed with appropriate professionals rather than treated as established fact. |
| **Safety and Quality Require Clear Information** | The quality and safety of TCIM-related activities depend not only on the products or practices themselves, but also on how information, uses, preparation methods, and potential effects are communicated to users. |
| **Technical Recommendations Must Be Adapted to the Local Context** | Technical recommendations need to take into account language, education level, cultural context, available resources, existing practices, and the realities of the organizations and communities involved. |
| **Community Engagement Strengthens Relevance** | Direct interaction with community members and users provides practical information that cannot be obtained from technical documentation alone. User feedback and community perspectives can help improve the presentation and practical use of TCIM-related products and information. |
| **Different Forms of Evidence Can Complement Each Other** | Documentary review, technical references, professional knowledge, community perspectives, process analysis, and simple user feedback can be combined to support practical decisions, particularly when formal evidence is limited. |
| **Simple Monitoring Can Support Continuous Improvement** | Simple approaches, such as tracking sales and collecting basic user feedback, can generate useful information for future learning and improvement without requiring complex monitoring systems. |
| **Technical Support Should Strengthen Existing Capacity** | Effective technical support should complement the knowledge and experience of health professionals and partner organizations rather than replace them. Technical review, discussion, adaptation, and shared validation can help strengthen local capacity and ownership. |
