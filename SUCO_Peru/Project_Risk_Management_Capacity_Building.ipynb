# Project Risk Management Capacity-Building

### Practical Tools, Risk Prioritization, and Risk-Informed Project Management

This project documents a capacity-building initiative conducted with two partner organizations in Peru to strengthen project risk management. Following an initial project management assessment, risk management was identified as a priority because neither organization had an established process for managing project risks.

The intervention combined the development of an Excel-based risk register, in-person training, practical exercises, and follow-up technical assistance. The approach focused on introducing a simple, structured method that project teams could apply during proposal development and project implementation.

This notebook presents the context and needs that led to the intervention, the design and implementation of the approach, the subsequent activities, and the lessons learned.

## 1.1-Context of the Assignment

The intervention was conducted with two partner organizations in Peru as part of broader efforts to strengthen project management practices. An initial organizational assessment had identified risk management as a priority for both organizations.

The assignment focused on developing a practical risk management tool and an associated training program for technical teams and project managers. The aim was to introduce a consistent approach that could support both the preparation of project proposals and the management of ongoing projects.

The intervention included training sessions delivered separately to each organization, followed by technical assistance to support the application of risk management in project-related situations.

## 1.2-Initial Needs Analysis

The initial assessment found that neither organization had an established practice for systematically managing project risks.

Several needs were identified across the risk management process:

- **Understanding risks:** Clarify what constitutes a risk and distinguish uncertain future events from problems that have already occurred.
- **Risk identification:** Help teams identify potential events or conditions that could affect project objectives.
- **Risk assessment:** Establish a structured way to assess the probability and potential impact of identified risks.
- **Risk prioritization:** Help teams determine which risks require the most attention.
- **Risk response and monitoring:** Support decisions about appropriate responses and establish basic arrangements for tracking risks over time.

These needs were relevant throughout the project life cycle. Risk analysis was important during the preparation of concept notes and proposals for funding partners, as well as during project implementation and monitoring. The intervention therefore needed to address the complete process while remaining practical and accessible to teams with different levels of digital literacy.

## 1.3-Strategy Design

The approach was designed around three principles: accessibility, practical learning, and application to real project management needs.

### Strategy Design Overview

| Component | Design Decision | Rationale |
|---|---|---|
| Risk management tool | Excel-based risk register | Use familiar software suited to the partners' digital literacy |
| Risk assessment method | Probability and impact ratings combined through a risk matrix | Provide a structured basis for assessing and prioritizing risks |
| Training format | Half-day, in-person session for each organization | Introduce the method and allow participants to practice it |
| Practical exercise | Fictional project to improve the income of small-scale rural producers | Apply the method to realistic implementation challenges |
| Follow-up support | Technical assistance for two project proposals and workshops supporting a project experiencing difficulties | Provide opportunities to apply the approach in different project contexts |

### 1.3.1 – Risk Management Method

The training defined a risk as an uncertain event that could affect project objectives. It introduced a standard formulation based on three elements:

                                                      Cause → Potential event → Consequence

The formulation helps clarify why a risk might occur, what could happen, and how the event could affect the project. For example, staff turnover could lead to a loss of technical capacity within a project team, potentially affecting activity implementation. Similarly, low participation among target populations could reduce attendance at activities and affect expected results.

Risk assessment was based on two dimensions:

- **Probability:** How likely is the risk to occur?
- **Impact or severity:** How significant would its consequences be?

The two ratings were combined to calculate a risk score:

                                                        Risk level = Probability × Impact

A color-coded matrix helped users interpret the resulting scores and identify priorities for action. The training also emphasized that risk management requires periodic review as project conditions change



### 1.3.2 – Risk Register Architecture

Excel was selected because it was familiar and appropriate to the partners' existing digital skills. The register was designed to capture the minimum information needed to document, assess, assign, and monitor risks. The design included predefined lists and an automatic risk-level calculation. This reduced manual work and supported consistent data entry without requiring additional software or a complex management system.

### 1.3.3 – Training and Practical Exercise Design

The training was structured around five components: objectives, key concepts, introduction to the risk register, a practical exercise, and a closing discussion with participant feedback. The practical exercise used a fictional two-year project intended to improve the income of small-scale rural producers through training, technical assistance, and the distribution of agricultural inputs.

The case presented three potential risk situations:

- **Staff continuity:** The technical team included four people, two of whom had one-year contracts despite the project's two-year duration.
- **Training schedules and participation:** Training was planned during periods of high agricultural activity, when previous participation had been variable.
- **Input procurement:** The project depended on a small number of local suppliers whose delivery times fluctuated.

These scenarios were selected to help participants practice risk formulation, assessment, prioritization, and response planning.

## 1.4-Tool Implementation and User Support

The intervention was implemented through the development of the Excel register, delivery of training sessions, and subsequent technical assistance.

### 1.4.1 – Development of the Excel Risk Register

The risk register was developed with the following fields:

| Field | Purpose |
|---|---|
| **Risk number** | Assign a reference number to each risk |
| **Project life-cycle phase** | Identify the relevant project phase |
| **Risk category** | Classify the risk |
| **Risk description** | Describe the potential event and its possible consequences |
| **Impact** | Record the assessed severity of the consequences |
| **Probability** | Record the assessed likelihood of occurrence |
| **Risk level** | Calculate the risk score automatically |
| **Person responsible for follow-up** | Identify who is responsible for monitoring the risk |
| **Risk status** | Track the current status of the risk |

The predefined **project life-cycle phases** were:
- Preliminary project development and initiation
- Planning
- Implementation
- Control
- Closure

The **risk categories** were:
- Environmental
- Financial
- Project management
- Governance
- Human resources
- Legal and regulatory
- Schedule
- Political
- Quality-related
- Technical

The **risk status options** were:
- Accepted
- Under analysis
- Under continuous monitoring
- Being treated
- Treated

The register was intentionally limited to these essential fields to keep its use manageable and support consistent risk documentation.



### 1.4.2 – Delivery of Training Sessions

A half-day, in-person training session was delivered to each partner organization. The sessions introduced the main concepts of risk management, the distinction between potential risks and existing problems, the formulation of risks, probability and impact assessment, and the use of the risk register. Participants were also introduced to the relationship between risk assessment, prioritization, response planning, and ongoing monitoring.

### 1.4.3 – Practical Application and Participant Feedback

During the practical exercise, participants worked with the fictional rural producer project and its implementation challenges. The exercise provided an opportunity to apply the method using the Excel register. The sessions concluded with a collective discussion about the benefits of risk management and a participant satisfaction assessment using a five-point scale, from very dissatisfied to very satisfied.

### 1.4.4 – Follow-Up Technical Assistance

Following the training, two technical assistance assignments supported the preparation of project proposals. These activities provided opportunities to apply risk management during project design. A further intervention supported a project experiencing implementation difficulties through workshops with participants. These workshops lasted approximately two hours and focused on applying risk management concepts in an operational context.

## 1.5-Post-Implementation Development and Future Enhancements

The follow-up activities provided opportunities to apply risk management during proposal preparation and in a project experiencing implementation difficulties. One important observation was that responding to major risks could require substantial resources. Teams recognized that it was not always feasible to implement every possible mitigation measure. This reinforced the importance of prioritizing risks, considering available resources, and determining which responses were realistic.

The available information supports the following assessment of the intervention and its potential next steps.

| Area | Observations | Further Work Required |
|---|---|---|
| **Risk prioritization** | The intervention reinforced the need to distinguish risks according to their importance and the resources needed to address them. | Continue helping teams prioritize risks using consistent assessment criteria. |
| **Risk tolerance** | Participants recognized that not every risk can be addressed to the same extent. | Explicitly consider risk tolerance and available resources when selecting responses to priority risks. |
| **Proposal development** | Two technical assistance assignments supported the application of risk management during proposal preparation. | Continue integrating risk analysis into concept notes and proposals where relevant. |
| **Project implementation** | A further intervention applied risk management concepts through workshops supporting a project experiencing difficulties. | Encourage regular review of risks, assigned responsibilities, and follow-up status as project conditions change. |
| **Continued use of the register** | The register provided a basic structure for documenting and monitoring risks. | Assess its continued use and identify any adjustments needed as teams gain experience. |

No longitudinal data were available to establish whether the register was adopted consistently over time or whether the intervention led to measurable improvements in project performance. Further follow-up would be required to assess sustained use and determine whether the tool needs to be adapted.

## 1.6-Key Lessons Learned

The intervention highlighted several lessons for designing practical capacity-building activities in project risk management.

| Lesson Learned | Application to Future Technical Assistance |
|---|---|
| **Start with the organization's needs and capacities** | Select methods and tools that reflect the team's existing skills, working practices, and available resources. |
| **Use clear and consistent risk formulation** | Help teams distinguish causes, uncertain events, and potential consequences when documenting risks. |
| **Distinguish risks from existing problems** | Focus risk assessment on uncertain events that could affect project objectives, while addressing problems that have already occurred through appropriate management actions. |
| **Make risk assessment systematic** | Use probability and impact ratings to establish a consistent basis for prioritization. |
| **Keep tools proportionate to operational needs** | Use familiar software, predefined lists, and automated calculations to limit the administrative burden. |
| **Combine training with practical exercises** | Use realistic project situations to help participants apply concepts and methods. |
| **Integrate risk management throughout the project life cycle** | Consider risks during proposal preparation, planning, implementation, monitoring, and closure. |
| **Account for resource constraints and risk tolerance** | Recognize that mitigation measures require resources and that teams need to make realistic choices about responses. |
| **Reinforce learning through application** | Complement initial training with technical assistance that supports the use of risk management in project-related situations. |

Overall, the intervention established a practical foundation for risk management by combining a simple tool, structured assessment criteria, and applied training. Its central lesson was that risk management needs to be both systematic and proportionate: teams must be able to identify and prioritize risks clearly while making realistic decisions about the responses they can implement and monitor.
