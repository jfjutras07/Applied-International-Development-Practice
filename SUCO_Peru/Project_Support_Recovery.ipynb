# Strengthening Project Management and Performance Monitoring in Partner Organizations in Peru

### Technical Assistance, Process Improvement, and Organizational Capacity Strengthening

This project presents four technical assistance interventions conducted with two partner organizations in Peru to strengthen project planning, clarify organizational responsibilities, address recurring implementation delays, and improve the definition of project results.

The interventions responded to practical management challenges identified within the organizations. They combined project management tools, participatory workshops, process analysis, root cause analysis, and indicator development to help teams better understand their challenges and structure their management practices.

Three interventions were conducted with a relatively mature partner organization, focusing on project scheduling, role clarification, and the analysis of recurring delays. A fourth intervention was conducted with another partner organization to strengthen its capacity to define and monitor project results through clearer indicators.

Rather than introducing complex management systems, the approach focused on practical methods and accessible tools that could be integrated into existing organizational practices. This notebook documents the needs addressed, the methods applied, and the main lessons emerging from these technical assistance activities.

## 1.1-Context of the Assignment

This work was carried out as part of technical assistance activities with partner organizations in Peru. The objective was to support organizational and project management practices by addressing specific challenges encountered during project implementation and performance monitoring. The interventions covered different but interconnected dimensions of project management. These included planning activities and deadlines, clarifying roles and responsibilities, understanding the causes of implementation delays, and defining indicators to assess project results.

The first three interventions were conducted with a relatively mature partner organization that had established project implementation experience but faced specific management challenges. The fourth intervention addressed the definition of project results with another partner organization. The approach was adapted to each situation rather than applying a single standardized solution. Each intervention began with a concrete management problem and focused on selecting appropriate methods to help the organization analyze the situation, structure information, and identify practical ways to improve its management practices.

## 1.2-Initial Needs Analysis

The technical assistance activities addressed four main needs:

- **Project delays**: Strengthening activity planning and scheduling to improve visibility over implementation sequences, deadlines, and dependencies.

- **Unclear responsibilities**: Establishing a clearer understanding of who was responsible for specific activities and decisions.

- **Recurring delays**: Moving beyond the observation that delays occurred toward a structured analysis of their underlying causes.

- **Unclear results**: Improving the definition of project results and establishing indicators that could support monitoring and assessment.

These needs required different methods. Planning and responsibility challenges called for tools to organize activities and clarify accountability. Recurring delays required a more analytical approach to understand contributing factors. Unclear results required collaborative work to translate project objectives into more specific and measurable indicators.

The interventions were therefore designed around the principle that effective project management requires both practical management tools and a clear understanding of the problems those tools are intended to address.

## 1.3-Strategy Design

The four interventions were designed around distinct project management challenges. The methods were selected according to the nature of each problem, the information available, and the practical requirements of the teams involved. The overall approach emphasized three principles: selecting methods suited to the problem, involving project team members in developing solutions, and keeping the resulting tools proportionate to operational needs.

### Strategy Design Overview

| Intervention | Selected Methods | Rationale |
|---|---|---|
| Climate risk management – La Libertad | Work Breakdown Structure (WBS) and Gantt chart | Structure project activities and establish a schedule with dates and dependencies |
| Food security – Northern Peru | RACI matrix and project organizational chart | Clarify responsibilities and represent project governance |
| Recycling and composting training – Northern Peru | Ishikawa diagram, weighted voting, Five Whys, and action plan | Explore potential causes, prioritize an issue for further analysis, and identify corrective measures |
| Community health promotion – High Andes | Four performance indicators and indicator sheets | Establish a manageable framework for monitoring activities, participation, satisfaction, and workload |

The RACI framework was selected to clarify responsibility assignments at the activity level. A project organizational chart would complement it by providing a visual overview of governance arrangements and relationships between stakeholders.

### 1.3.3 – Root-Cause Analysis

The Ishikawa diagram was selected to organize the exploration of potential causes using the six traditional categories known as the 6 Ms: Manpower, Methods, Machines, Materials, Measurement, and Mother Nature (environment). Weighted voting would support collective prioritization, and the Five Whys would allow further examination of the selected cause.



Pareto analysis was not selected because quantitative data were unavailable to compare the frequency or relative contribution of different causes. It could be considered later if suitable data become available.

### 1.3.4 – Performance Monitoring

The monitoring framework was deliberately limited to four indicators. Two would formalize existing measures, while two additional indicators would address participant satisfaction and estimated workload. This approach aimed to provide useful management information without creating an unnecessarily demanding reporting system.

## 1.4-Tool Implementation and User Support

The selected methods were translated into practical tools through document development, collaborative workshops, and structured analysis with the project teams.

### 1.4.1 – Project Planning Tools

**Project:** Climate risk management in the La Libertad region.

The initial project documentation was reviewed and used as the starting point for a workshop with project team members. Their input helped expand and refine the activity structure, resulting in a more complete Work Breakdown Structure (WBS).

During a second workshop, the activities were transferred into a Gantt chart incorporating start dates, end dates, and dependencies.

A person responsible for updating the schedule was identified, and the planning tools were integrated into the project's monitoring committee.

### 1.4.2 – Responsibility and Governance Tools

**Project:** Food security in northern Peru.

Preliminary interviews with project team members helped identify the activities and stakeholders to include in the RACI matrix. The level of detail was considered carefully: broad activity descriptions can leave responsibilities unclear, while excessive fragmentation can make the matrix difficult to maintain.

During the workshop, participants reviewed the activities and assigned responsibilities using the RACI framework:

- **R – Responsible:** The person or people carrying out the activity.
- **A – Accountable:** The person ultimately accountable for the activity's completion or outcome.
- **C – Consulted:** People who provide input through active consultation.
- **I – Informed:** People who need to receive relevant information.

The workshop addressed practical rules for using the matrix, including assigning a single accountable person to each activity and distinguishing consultation from information sharing.

A project organizational chart was also developed to illustrate the governance structure and relationships between the different actors.

### 1.4.3 – Root-Cause Analysis and Action Planning

**Project:** Recycling and composting training in northern Peru.

A workshop used the Ishikawa diagram to explore potential causes of delays in training activities for the local population. Participants examined the six cause categories and used weighted voting to prioritize the issues identified. Insufficient initial communication was selected for further analysis.

The Five Whys technique was then applied:

1. **Why were training activities delayed?** Attendance was not confirmed sufficiently in advance.
2. **Why was attendance not confirmed in advance?** Participants were not contacted early enough to confirm their participation.
3. **Why was early communication insufficient?** The initial communication arrangements did not ensure that confirmations would be obtained before the training.
4. **Why did the arrangements not ensure confirmation?** The follow-up process was not sufficiently structured.
5. **Why was the follow-up process not structured?** Systematic attendance confirmation had not been adequately incorporated into the preparation process.

A brainstorming session followed the analysis. The proposed solutions were consolidated into a short action plan focused on improving initial communication and confirming attendance before training activities.

### 1.4.4 – Performance Monitoring Tools

**Project:** Community health promotion in the high Andes.

Four indicators were formalized to support project monitoring. Two documented measures already being collected informally, while two additional indicators addressed participant satisfaction and the estimated working hours required for project activities.

| Indicator | Definition | Data Source |
|---|---|---|
| Number of community health promotion activities delivered | Total number of activities conducted during the reporting period | Activity register |
| Number of people reached | Number of participants attending activities, distinguishing unique individuals from repeat attendance where feasible | Attendance sheets |
| Participant satisfaction rate | Percentage of respondents reporting satisfaction, based on a standardized feedback question | Short participant feedback form |
| Estimated total effort in hours | Estimated working hours required to prepare, deliver, and follow up on project activities | Activity plan and staff time estimates |

Collaborative workshops helped clarify the indicators and establish a shared understanding of their definitions and intended use. Indicator sheets were prepared to document definitions, calculation methods where applicable, and data sources.

## 1.6-Key Lessons Learned

The four technical assistance interventions highlighted several practical lessons for supporting project management in partner organizations.

| Lesson Learned | Application to Future Technical Assistance |
|---|---|
| **Adapt tools to the operational context** | Select methods and tools that address the specific management challenge while remaining proportionate to the team's resources and capacity. |
| **Involve project teams in developing solutions** | Use interviews and collaborative workshops to incorporate team knowledge, clarify responsibilities, and encourage ownership of the resulting tools. |
| **Translate analysis into concrete actions** | Link problem analysis to an action plan with clearly defined follow-up activities rather than stopping at the identification of potential causes. |
| **Distinguish immediate improvements from sustained results** | Monitor changes over time before concluding that a corrective action has resolved a problem. Where several causes may contribute to an issue, investigate them progressively. |
| **Keep performance monitoring manageable** | Start with a limited number of clearly defined indicators and practical data sources. Expand the framework only when additional information is useful and feasible to collect. |
| **Plan for continued use from the outset** | Establish responsibility for updating tools and connect them to existing management routines, such as project monitoring committees. |
| **Allow for progressive refinement** | Treat management tools as working documents that can be adjusted as projects evolve and further information becomes available. |
