# Designing and Implementing Fundraising Strategies for Partner Organizations in Peru
### Analysis, Strategy Design, and Implementation

This project presents the development of fundraising strategies for two partner organizations in Peru, designed to strengthen their capacity to identify, prioritize, and develop funding opportunities in alignment with their institutional objectives.

The strategies provide a practical framework for organizing fundraising efforts, diversifying funding sources, developing relationships with potential funders and partners, and connecting funding opportunities with organizational priorities. They are supported by simple tools for stakeholder analysis, project prioritization, funding opportunity tracking, responsibility allocation, and concept note development.

The objective was not to introduce costly or complex fundraising management systems, but to demonstrate how clear processes, well-designed tools, and effective organization of information can significantly strengthen resource mobilization practices.

This notebook documents the analysis, strategy design, and practical tools developed to support fundraising activities. It illustrates how a structured and accessible approach can help organizations better organize their efforts, make more informed decisions about funding opportunities, and progressively strengthen their fundraising capacity.

## 1.1-Context of the Assignment

This work was carried out as part of my assignment with SUCO in Peru, where I contributed to project management, organizational development, and resource mobilization activities with partner organizations.

Within the Peruvian operational context, partner organizations regularly face the challenge of identifying and securing financial resources to maintain and expand their development initiatives. Funding opportunities may originate from bilateral and multilateral cooperation agencies, international funds, philanthropic foundations, financial institutions, private-sector organizations, universities, and other potential partners.

Both organizations had established experience in implementing development projects and managing relationships with different types of funding partners. However, fundraising activities were not yet supported by a comprehensive medium-term framework to systematically organize funding opportunities, prioritize initiatives, develop partnerships, and coordinate resource mobilization efforts.

At the same time, the funding environment was evolving, with changing donor priorities, increasing requirements for project proposals, and a growing need for organizations to diversify their funding sources. These conditions created a need to strengthen internal processes for anticipating opportunities and connecting them more systematically with institutional priorities.

Within this context, my mandate included supporting the development of fundraising strategies for two partner organizations. The work focused on establishing practical frameworks that could help each organization organize its fundraising activities, identify relevant funding opportunities, prioritize potential initiatives, and strengthen relationships with funders and other partners.

The approach was intentionally designed around accessible tools and existing organizational capacities rather than specialized or costly fundraising management systems. The objective was to develop practical mechanisms that could be integrated into existing planning and project development practices and maintained by the organizations over time.


## 1.2-Initial Needs Analysis

The initial needs analysis was conducted through discussions with staff and partner organizations and a review of existing fundraising and project development practices.

The analysis showed that no formal fundraising strategy or structured resource mobilization process was in place. Funding opportunities were generally pursued as they became available, often requiring proposals to be developed under significant time constraints and without a standardized methodology or format.

The organizations also lacked mechanisms for systematic funding diversification, forward planning, and project prioritization. In some cases, portfolios of potential initiatives existed, but projects had not been assessed using common criteria to determine which initiatives should be developed and pursued.

The analysis therefore highlighted the need for a practical framework capable of moving fundraising from a predominantly reactive approach toward a more **planned, structured, and proactive process**. The main needs identified were:

* Establishing a formal fundraising strategy aligned with institutional priorities
* Creating a structured system for tracking and assessing funding opportunities
* Introducing consistent criteria for project prioritization
* Supporting greater diversification and forward planning
* Standardizing the early development of project ideas
* Clarifying roles and responsibilities within the fundraising process

Given the organizations' familiarity with standard office software, the solution focused on accessible and maintainable tools developed in **Word and Excel**, including automated Excel functions and macros where appropriate.


## 1.3-Strategy Design

### Initial Design Phase

Following the initial needs analysis, the development effort focused on translating the identified fundraising gaps into a practical and integrated framework. The objective was to establish a clear process connecting funding opportunity identification, project prioritization, proposal development, and follow-up.

The design was based on a simple principle: **fundraising should be organized as a continuous process rather than as a series of last-minute responses to available funding opportunities**.

The resulting approach combined a medium-term fundraising strategy with a set of practical tools designed to support its implementation. The tools were developed using Word and Excel, with automated functions and macros incorporated where they could facilitate data management, reduce repetitive tasks, and improve usability.

### Strategy and Tool Architecture

The resulting system was designed around a continuous fundraising cycle rather than a single tool. The **fundraising strategy** establishes the overall institutional direction, while the supporting tools provide practical mechanisms for translating this direction into day-to-day activities.

The strategy establishes the main priorities for fundraising, including thematic areas, funding diversification, partnership development, project prioritization, governance, and monitoring. It provides the common framework within which the different tools are used.



The process begins with the identification and monitoring of funding opportunities. Relevant opportunities are recorded in the **Funding Opportunities Directory**, which centralizes information such as funders, funding modalities, available amounts, deadlines, thematic areas, application links, and proposal status. The tool also uses visual indicators to facilitate the monitoring of approaching deadlines and the status of opportunities.



Potential projects are assessed separately through the **Project Prioritization Matrix**. The matrix uses eight criteria covering strategic alignment, contribution to development results, urgency, impact and equity, resource requirements, implementation complexity, risk and uncertainty, and dependencies. Each criterion uses an anchored **1–10 scale**, providing a common basis for comparing potential initiatives.

The criteria were designed to consider both the potential value of an initiative and the conditions required for its implementation. This allows project discussions to move beyond the simple availability of funding and consider whether an initiative is strategically relevant and realistically implementable.

The **Stakeholder Analysis** tool supports the identification and characterization of potential funders and partners. It provides a structured way to assess stakeholders according to their role, level of engagement, influence, potential concerns, motivations, and relevance to the organization. This analysis supports the development of appropriate engagement strategies and helps organizations determine how relationships with potential funders and partners should be approached and managed.

The **RACI Matrix** was developed to clarify responsibilities throughout the fundraising process. It covers activities ranging from identifying and monitoring funding opportunities to prioritizing projects, matching projects with funding opportunities, developing and submitting proposals, and monitoring results and lessons learned.

Once a project and funding opportunity are considered relevant, the **Concept Note Template** provides a standardized structure for developing the initial project concept before progressing toward a complete proposal. This helps reduce the need to start each proposal from scratch and provides a common format for documenting project ideas.



An additional AI-assisted workflow was developed to support the adaptation of concept notes to specific funding opportunities. The process begins with the original concept note and the official documentation of the potential funder. It then separates the analysis of alignment and gaps from the subsequent adaptation of the project. Specific controls were included to prevent unsupported information from being introduced and to ensure that substantive modifications remain subject to organizational validation.

Together, these components create a structured process linking **institutional priorities → project prioritization → funding opportunities → project–funder alignment → concept development → proposal adaptation → submission and learning**.

The overall design intentionally relied on familiar and accessible technologies rather than specialized fundraising management software. This approach allowed the organizations to maintain and adapt the tools internally while establishing a more systematic approach to resource mobilization.

## 1.4-Tool Implementation and User Support

The implementation phase was structured around a series of workshops with key participants from each partner organization. Between **5 and 10 participants per organization** were involved in the co-creation and validation of the different components of the fundraising framework.


Rather than developing the tools independently and transferring them at the end of the assignment, the process allowed participants to review the proposed approaches, provide feedback, identify adjustments, and validate the tools against their organizational realities.

The workshops covered the different components of the fundraising system, including the fundraising strategy, funding opportunities directory, project prioritization approach, stakeholder analysis, roles and responsibilities, and concept note development. This iterative process helped ensure that the tools were understandable, relevant to the organizations' needs, and compatible with their existing practices.

Following the development and validation of the tools, technical support was provided to help the organizations use the fundraising process as an integrated sequence rather than as a collection of independent tools. Particular attention was given to the management and updating of the **Funding Opportunities Directory**, including the organization of information, monitoring of opportunities, and use of its different functions.

Technical assistance was also provided for minor adjustments required during implementation and for practical questions arising from the use of the tools. This support was intended to progressively build the organizations' capacity to use and maintain the framework independently.

The implementation therefore followed a progressive approach combining **co-creation, validation, guided use, technical support, and preparation for knowledge transfer**. In accordance with the RACI Matrix, a designated person within each organization is expected to assume responsibility for the continued management of the fundraising tools. Additional accompaniment is planned before the final transfer to ensure that the designated staff have the necessary understanding and practical experience to maintain the system.


## 1.5-Post-Implementation Development and Future Enhancements

Following the initial implementation, the fundraising framework is expected to evolve through continued use, technical support, and progressive internal ownership within each partner organization.

A final accompaniment phase is planned before the complete transfer of responsibility to a designated person within each organization, in accordance with the roles defined in the RACI Matrix. This phase will focus on consolidating practical knowledge, resolving minor technical issues, and ensuring that the organizations are comfortable managing the tools independently.

The Funding Opportunities Directory is also designed to evolve as organizations accumulate information on funders, opportunities, applications, and results. Its dashboard can progressively support management follow-up and accountability by providing a consolidated view of fundraising activities and results.



The tools can also be refined over time as organizations gain experience using them. Adjustments to the directory, prioritization matrix, reporting mechanisms, and other components can be introduced based on practical experience and emerging organizational needs.

Future development should therefore focus not on increasing the technical complexity of the system, but on progressively strengthening its use, quality of information, internal ownership, and integration into regular organizational management practices.

The framework is intended to become a living management system that can be updated as funding environments, institutional priorities, partnerships, and organizational capacities evolve.

## 1.6-Lessons Learned

Several lessons emerged from the development and implementation of the fundraising framework.

**Co-Creation Improves Relevance and Ownership**

The involvement of key participants from each organization throughout the design and validation process was essential. Rather than developing the tools independently, the workshops created opportunities to test assumptions, incorporate organizational knowledge, and adjust the tools to actual working practices.

This approach also contributed to ownership. Staff were not only introduced to completed tools but participated in shaping how the tools would be used within their organization.

**Fundraising Requires an Integrated Process**

The project highlighted that fundraising tools are most useful when they are connected through a common process. A funding directory alone does not create a fundraising strategy, just as a project prioritization matrix does not determine which opportunities an organization should pursue.

The value of the framework comes from connecting the different components: strategy, projects, funding opportunities, stakeholders, responsibilities, concept development, and proposal adaptation.

**Simple Tools Can Support More Structured Management**

The implementation demonstrated that organizations do not necessarily require specialized fundraising management software to establish a more systematic approach. Well-designed Word and Excel tools, combined with appropriate automation, can provide a practical infrastructure for organizations that already rely on these technologies.

The emphasis should therefore be placed on process design, information organization, usability, and ownership, rather than on technological complexity.

**Technical Simplicity Does Not Mean Organizational Simplicity**

Although the tools themselves were designed to remain accessible, their effective use requires an understanding of the complete fundraising sequence. Technical support was therefore still necessary during implementation, particularly for managing the funding directory and making minor adjustments.

This highlighted the importance of combining tool delivery with guided implementation and knowledge transfer.

**Monitoring and Accountability Should Be Built into the System**

The inclusion of a dashboard within the Funding Opportunities Directory illustrated the value of integrating management follow-up into the fundraising process itself. Rather than treating reporting as a separate administrative activity, relevant information can be structured from the beginning so that it supports both operational monitoring and organizational accountability.

**Sustainability Depends on Internal Ownership**

Finally, the long-term value of the framework will depend on its continued use and maintenance by the organizations themselves. The planned transfer of responsibilities to designated staff, supported by the RACI Matrix and additional technical accompaniment, is therefore an important component of implementation rather than an activity occurring after the project is completed.

Overall, the experience reinforced the importance of combining co-creation, simple technology, structured processes, and progressive knowledge transfer when introducing new management practices within partner organizations.
