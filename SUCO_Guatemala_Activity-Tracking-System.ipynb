# Designing and Implementing an Excel-Based Tool for Funding and Activity Tracking
## Analysis, Tool Design, and Implementation

This project presents a simple tracking tool designed to monitor funding sources and their associated activities. The tool allows users to register funding sources, link them to one or more activities, track their status over time, close completed activities, and consult a dashboard summarizing the overall situation.

The focus of this notebook is not on advanced analytics or predictive modeling, but on structuring information in a clear and consistent way. The proposed approach relies on basic data entry forms, structured tables, and a dashboard built from these tables to support day-to-day monitoring and reporting.

This notebook documents the logic behind the data structure, the relationships between funding sources and activities, and the basic indicators displayed in the dashboard. It illustrates how a lightweight system can support funding and activity follow-up without requiring complex information systems.

## 1.3-Tool Design

### Initial Consolidation Phase

With the information gathered during the initial analysis, I began the design of the tool. At the time, multiple Excel-based tools already existed for activity and project follow-up. However, these tools were largely unused by partner organizations and were maintained as separate and disconnected files.

These existing tools included a simple directory of funding sources, as well as various Excel files reflecting different aspects of project management. Some focused on project charters, others on risk matrices, budget tracking (using several different formats), or activity indicator tracking. In addition, separate sheets were used to document objectives.

A key challenge identified during this phase was the lack of consistency across these tools. Different definitions were used for similar concepts, particularly for activities that were not formally structured as projects. This resulted in grey areas in terminology, scope, and follow-up practices, making consolidated tracking difficult.

Given the level of digital literacy among partner organizations and the tools currently available to them, the choice of Excel was not optional but necessary. Excel represented the most appropriate technological solution, both in terms of accessibility and change management. It allowed the tool to build on familiar practices while introducing greater structure and consistency without creating additional barriers to adoption.

The initial design effort therefore focused on consolidating essential information into a single, coherent Excel-based structure. The objective was not to reproduce all existing tools, but to identify and retain the elements necessary for funding and activity tracking, while reducing fragmentation and improving clarity for partners and SUCO staff.

### Identified Needs and Design Trade-Offs

During consultations with SUCO and partner organizations, multiple needs were identified. These included the need for a centralized directory of funding sources, a tool to support project follow-up, and the integration of operational elements required for accountability, such as deliverables and the number of people reached.

While these needs were legitimate and reflected real operational requirements, addressing all of them within a single tool would have significantly increased complexity. From a design and change management perspective, this level of complexity risked creating a gap between the tool and its intended users, potentially limiting adoption and effective use.

A key design decision was therefore required. In order to facilitate both the design of the tool and its later adoption by partners, prioritization was necessary. Rather than attempting to cover all project management and reporting dimensions at once, the focus was placed on a limited and coherent scope aligned with immediate tracking needs.

This choice aimed to reduce barriers to use and ensure that the tool remained accessible and practical. The underlying assumption was that an overly complex solution would be counterproductive and could ultimately undermine the objectives of the initiative.

| Area                                  | Identified Needs                                                                                                                 | Design Decision                                                                                                                      |
| ------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------ |
| **Funding Source Directory**          | Centralized repository including donor organization, funding source, deadline, and funding amount to support follow-up over time | **Fully integrated** as a core component of the tool to ensure centralization and temporal tracking                                  |
| **Project Follow-up**                 | Basic project management elements, mainly timelines and budgets, without adding unnecessary complexity                           | **Partially integrated**, limited to essential fields to keep the tool usable                                                        |
| **Projects vs. Operations**           | Ability to distinguish between projects and operational activities                                                               | **Integrated at a basic level** to enable follow-up, while allowing future refinement as capacities and the tool evolve              |
| **Objectives**                        | Ability to link objectives to activities                                                                                         | **Excluded** due to ongoing strategic planning processes and limited training on SMART objectives; qualitative notes remain possible |
| **Risk Management**                   | Risk identification and monitoring                                                                                               | **Excluded** to reduce scope and avoid overlap with existing tools                                                                   |
| **Other Project Management Elements** | Decisions, governance, and additional follow-up components                                                                       | **Excluded** to maintain a lightweight and focused design                                                                            |
| **Deliverables**                      | Categorization of deliverables (e.g., infrastructure, workshops, organized events, contracting)                                  | **Integrated** to support operational and reporting needs                                                                            |

### Tool Structure and Architecture

User stories and use cases were identified during the analysis phase to clarify the intended uses of the tool and the needs of its future users. However, they are not presented explicitly in this notebook in order to keep the documentation concise and avoid overloading the presentation.

Instead, the focus is placed on presenting the tool itself (its structure, logic, and key components) as these elements directly reflect the identified use cases and operational needs.

**Presentation**

The tool includes a dedicated Presentation sheet that provides an overview of its structure and main components. This sheet describes the purpose of each worksheet and explains how they relate to one another within the overall tracking logic.

**Funding Source Directory**

For the funding source directory, the tool includes a dedicated data entry form used to capture the required information. Once completed, the information entered through the form is automatically recorded in a structured reference table.

Conditional formatting is applied to the table to provide a clear visual indication of upcoming deadlines, helping users anticipate funding offer expirations over time.

In addition, all forms in the tool include forcing functions to ensure that required fields are completed correctly. Clear instructions are provided directly within the forms to guide users and improve data quality.

**Activity Entry Form**

The tool includes a dedicated form to register activities, whether structured as projects or as operational activities. Each activity is identified by a title and associated thematic areas to support classification and follow-up.

An initial budget is entered using a limited set of predefined categories to support basic budget follow-up and quality management, without introducing unnecessary complexity at this stage. The budget categories include human resources, activities, materials and assets, transportation, food, management, and other costs.

Since funding sources may be provided in different currencies, the exchange rate in effect at the time of activity creation is used for initial calculations. This exchange rate can be updated during follow-up to reflect changes over time.

For timeline tracking, each activity includes a start date and an end date, enabling basic monitoring of planned durations.

All information described above is calculated automatically within the tool. Users are only required to complete the white input fields, while all other fields are populated through formulas to ensure consistency and reduce data entry errors.

**Activity Follow-up and Monitoring**

Activity follow-up is conducted through a dedicated monitoring interface. Users select an activity from a drop-down menu, which automatically displays the date of the most recent follow-up. When a new follow-up date is entered, the current percentage of physical progress is updated accordingly.

Budget follow-up is carried out by completing the relevant expense fields. Based on these inputs, the tool automatically calculates the percentage of the budget used, allowing users to monitor financial progress alongside activity implementation.

The follow-up process also includes a simplified mechanism to record completed deliverables. Deliverables can be selected from predefined categories such as workshops, trainings, events, tools, documents, asset purchases, and infrastructure. When applicable, participant information can also be recorded.

This approach enables tracking of key activity outputs commonly required by most donors, while deliberately avoiding excessive complexity during the first phase of implementation. The focus remains on practicality, usability, and consistent follow-up.

**Activity Closure**

Activity closure represents the final step of the follow-up process and is carried out once all planned elements have been completed or when a decision is made to stop the activity.

To close an activity, users only need to select the activity name and enter the closure date. Once this information is provided, all related activity data—including percentage of progress, budget information, completed deliverables, and participant data—is automatically displayed for verification.

The closure process also allows the optional entry of qualitative information, such as achievements, challenges, and lessons learned. This section can be completed when relevant, providing additional context without being mandatory.

Closing an activity removes it from active follow-up menus, allowing a clear distinction between ongoing and completed activities. This mechanism supports a cleaner monitoring process and helps maintain an up-to-date overview of current operations.

**Dashboard Overview**

The dashboard is designed to be simple and user-friendly, organized into two main sections: **Ongoing Activities** and **Completed Activities**. Users can filter the display by activity or by funding source to focus on specific items of interest.

For each activity, the dashboard presents the current percentage of physical progress and budget utilization, along with completed deliverables and participant information. This consolidated view allows users to quickly assess the status of activities and funding.

The dashboard can be used in various contexts, including follow-up meetings, coordination with partners, and report preparation. Its design prioritizes accessibility and clarity, ensuring that key indicators are available at a glance without introducing unnecessary complexity.

## 1.5-Post-Implementation Development and Future Enhancements

Following the initial implementation and adoption of the tool, a second phase is planned to build on the foundations established during Phase 1. This phase is intentionally positioned after the stabilization of initial use, ensuring that future enhancements respond to real operational needs and evolving capacities.

| Planned Area | Description | Intended Objective |
|-------------|-------------|-------------------|
| **Extended Capacity Building** | Additional colleagues working within more partner organizations will be trained once the tool is fully integrated into regular operations. | Ensure sustained adoption and reinforce shared project and activity follow-up practices. |
| **Progressive Link with Strategic Planning** | Selected activity-level objectives may be linked to ongoing strategic planning exercises being conducted with partner organizations. | Improve alignment between operational monitoring and medium-term strategic priorities. |
| **Initial Risk Management Integration** | A basic and optional risk identification component will be introduced, adapted to partners’ current levels of capacity and maturity. | Foster early risk awareness without increasing complexity or duplication. |
| **SUCO Internal Prioritization Tool** | A complementary tool will be developed for SUCO to support the prioritization of project proposals and ensure internal follow-up of projects and operational activities. | Strengthen internal decision-making, coordination, and resource allocation. |

This second phase is conceived as an incremental evolution rather than a redesign. Enhancements will be introduced progressively, based on observed use, user feedback, and organizational readiness, in order to preserve usability and long-term engagement.

As part of this phase, the system and accompanying training materials will also be translated into French, enabling the tool to be deployed in other regions where SUCO operates and where similar project management and monitoring needs have been identified.

## 1.6-Key Lessons Learned

A key lesson from this initiative is that simplicity is not a limitation but a condition for adoption. In contexts characterized by diverse partners, varying levels of digital literacy, and multiple operational constraints, a lightweight and accessible tool proved more effective than a comprehensive but complex system. Early prioritization and clear design trade-offs helped reduce resistance to change and supported consistent use. The experience also highlighted the importance of combining tools with capacity building, practical training, and ongoing support. Finally, anchoring tool development in real operational needs and iterative feedback emerged as a critical factor for sustainability and long-term value.
